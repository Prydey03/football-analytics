```python
shooting = fbref.read_player_season_stats(stat_type='shooting')
misc = fbref.read_player_season_stats(stat_type='misc')

shooting.columns = ['_'.join(col).strip('_') for col in shooting.columns]
misc.columns = ['_'.join(col).strip('_') for col in misc.columns]

print(shooting.columns.tolist())
print(misc.columns.tolist())
```

```python
merged = stats.merge(shooting, on=['league', 'season', 'team', 'player'], suffixes=('', '_shooting'))
merged = merged.merge(misc, on=['league', 'season', 'team', 'player'], suffixes=('', '_misc'))
print(merged.shape)
merged.head()
```

```python
print(f"Original stats rows: {len(stats)}")
print(f"Merged rows: {len(merged)}")
```

```python
model_data = merged[merged['Playing Time_90s'] >= 10].copy()
model_data = model_data.reset_index()
print(f"Players with 10+ full matches worth of minutes: {len(model_data)}")
```

```python
feature_cols = [
    'Standard_Gls', 'Standard_Sh', 'Standard_SoT%', 'Standard_G/Sh',
    'Performance_Ast', 'Performance_Crs', 'Performance_TklW',
    'Performance_Int', 'Performance_Fls', 'Performance_Fld'
]
```

```python
per_90_needed = ['Standard_Gls', 'Standard_Sh', 'Performance_Ast', 'Performance_Crs',
                  'Performance_TklW', 'Performance_Int', 'Performance_Fls', 'Performance_Fld']

for col in per_90_needed:
    model_data[col + '_p90'] = model_data[col] / model_data['Playing Time_90s']
```

```python
from sklearn.preprocessing import StandardScaler

final_features = ['Standard_Gls_p90', 'Standard_Sh_p90', 'Standard_SoT%', 'Standard_G/Sh',
                   'Performance_Ast_p90', 'Performance_Crs_p90', 'Performance_TklW_p90',
                   'Performance_Int_p90', 'Performance_Fls_p90', 'Performance_Fld_p90']

X = model_data[final_features].fillna(0)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.neighbors import NearestNeighbors

nn_model = NearestNeighbors(n_neighbors=6, metric='cosine')
nn_model.fit(X_scaled)
```

```python
model_data.columns.tolist()
```

```python
model_data = model_data.reset_index()
model_data.columns.tolist()
```

```python
def find_similar_players(player_name, n=5):
    idx = model_data[model_data['player'] == player_name].index
    if len(idx) == 0:
        return f"Player '{player_name}' not found."
    
    idx = idx[0]
    row_position = model_data.index.get_loc(idx)
    
    distances, indices = nn_model.kneighbors([X_scaled[row_position]])
    
    similar_players = model_data.iloc[indices[0][1:n+1]][['player', 'team', 'pos'] + final_features]
    return similar_players

find_similar_players('Bukayo Saka')
```

## Conclusion

Built a player-similarity engine using 2024-25 Premier League data from 
FBref (322 players with 10+ full matches played), combining standard, 
shooting, and miscellaneous stat categories into 10 per-90 features 
covering attacking output, shot quality, and defensive contribution.

**Method:** features were standardized (z-scores) and compared using 
cosine similarity via k-nearest-neighbors, returning the 5 most similar 
players to any given input.

**Validation:** the model performs strongly for attacking players — 
Bukayo Saka's closest matches (Trossard, Sávio, Eze, Traoré, Doku) are all 
recognizable wide-attacker/dribbler profiles, and Erling Haaland's matches 
(Isak, Mateta, Wood, Wissa, Welbeck) are all high-efficiency penalty-box 
finishers. Both results align with genuine football intuition.

**Limitation:** defensive midfielder comparisons are less reliable. 
Declan Rice's closest matches skewed toward creative attacking midfielders 
rather than genuine ball-winners, likely because only two defensive 
metrics (tackles won, interceptions per 90) were available through 
FBref's data via `soccerdata`, versus a much richer set of attacking 
features. I confirmed this is a genuine data ceiling — `soccerdata`'s 
FBref wrapper does not expose a dedicated passing or defensive stat 
category for player-season stats (checked via the library's own 
documentation) — rather than an oversight in feature selection.
